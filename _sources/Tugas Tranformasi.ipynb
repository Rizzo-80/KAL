# Tugas Matriks Transformasi

Diketahui:

Dari grafik, diperoleh koordinat titik:

- A = (2, 3)  
- B = (2, 1)  
- C = (4, 1)  
- D = (2, 4)  
- E = (2, 0)  
- F = (4, 0)  


##  Rumus Translasi
Transformasi translasi dinyatakan dengan:

$$
T =
\begin{bmatrix}
1 & 0 & a \\
0 & 1 & b \\
0 & 0 & 1
\end{bmatrix}
$$

Keterangan:
- $a$ = perubahan pada sumbu x  
- $b$ = perubahan pada sumbu y  


##  1. Transformasi A → B

Langkah 1: Perubahan koordinat
$$
A(2,3) \rightarrow B(2,1)
$$

$$
\Delta x = 2 - 2 = 0
$$

$$
\Delta y = 1 - 3 = -2
$$

Langkah 2: Matriks transformasi
$$
T_{A \to B} =
\begin{bmatrix}
1 & 0 & 0 \\
0 & 1 & -2 \\
0 & 0 & 1
\end{bmatrix}
$$


## 2. Transformasi B → C
 Langkah 1:
$$
B(2,1) \rightarrow C(4,1)
$$

$$
\Delta x = 4 - 2 = 2
$$

$$
\Delta y = 1 - 1 = 0
$$

Langkah 2:
$$
T_{B \to C} =
\begin{bmatrix}
1 & 0 & 2 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}
$$


## 3. Transformasi D → E

 Langkah 1:
$$
D(2,4) \rightarrow E(2,0)
$$

$$
\Delta x = 2 - 2 = 0
$$

$$
\Delta y = 0 - 4 = -4
$$

Langkah 2:
$$
T_{D \to E} =
\begin{bmatrix}
1 & 0 & 0 \\
0 & 1 & -4 \\
0 & 0 & 1
\end{bmatrix}
$$


##  4. Transformasi E → F

 Langkah 1:
$$
E(2,0) \rightarrow F(4,0)
$$

$$
\Delta x = 4 - 2 = 2
$$

$$
\Delta y = 0 - 0 = 0
$$

Langkah 2:
$$
T_{E \to F} =
\begin{bmatrix}
1 & 0 & 2 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

---

Semua transformasi merupakan **translasi**:
- A → B : turun 2 satuan  
- B → C : ke kanan 2 satuan  
- D → E : turun 4 satuan  
- E → F : ke kanan 2 satuan

## Matriks Transformasi Refleksi

 Diketahui
- A = (2,3) → B = (2,1)  
- B = (2,1) → C = (4,1)  
- D = (2,4) → E = (2,0)  
- E = (2,0) → F = (4,0)  

---

## Rumus Refleksi
Refleksi terhadap garis y = k
$$
T =
\begin{bmatrix}
1 & 0 & 0 \\
0 & -1 & 2k \\
0 & 0 & 1
\end{bmatrix}
$$
 Refleksi terhadap garis x = h
$$
T =
\begin{bmatrix}
-1 & 0 & 2h \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}
$$
 1. A → B (refleksi terhadap y = 2)

$$
T_{A \to B} =
\begin{bmatrix}
1 & 0 & 0 \\
0 & -1 & 4 \\
0 & 0 & 1
\end{bmatrix}
$$

 2. B → C (refleksi terhadap x = 3)

$$
T_{B \to C} =
\begin{bmatrix}
-1 & 0 & 6 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}
$$
 3. D → E (refleksi terhadap y = 2)

$$
T_{D \to E} =
\begin{bmatrix}
1 & 0 & 0 \\
0 & -1 & 4 \\
0 & 0 & 1
\end{bmatrix}
$$
 4. E → F (refleksi terhadap x = 3)

$$
T_{E \to F} =
\begin{bmatrix}
-1 & 0 & 6 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}
$$
## Kesimpulan
Transformasi yang terjadi adalah **refleksi (pencerminan)**:
- Terhadap garis y = 2  
- Terhadap garis x = 3  

Hasil akhir:
- (2,1), (4,1), (2,0), (4,0)
