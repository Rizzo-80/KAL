# Materi 5 Transformasi

Dalam aljabar linear, transformasi merupakan jembatan yang menghubungkan antara teori matriks dengan representasi grafis atau visual. Halaman ini akan membahas definisi, konsep dasar, serta syarat-syarat sebuah transformasi dikatakan linear.

---

## 1. Apa itu Transformasi?

Secara sederhana, **Transformasi** adalah sebuah aturan atau fungsi yang memetakan (mengubah) suatu vektor dari satu ruang dimensi (domain) ke ruang dimensi yang lain (kodomain). 

Jika kita memiliki fungsi transformasi $T$, kita bisa menuliskannya sebagai:
$$T: \mathbb{R}^n \rightarrow \mathbb{R}^m$$

Artinya, fungsi $T$ mengambil sebuah vektor masukan yang berada di ruang dimensi $n$ ($\mathbb{R}^n$), lalu memprosesnya dan menghasilkan vektor baru yang berada di ruang dimensi $m$ ($\mathbb{R}^m$).



### Analogi Sederhana
Bayangkan transformasi seperti sebuah **mesin cetak atau filter foto**:
* **Input:** Vektor awal berupa koordinat sebuah titik (misal posisi gambar persegi).
* **Proses (Transformasi):** Mesin melakukan operasi seperti menggeser, memutar (rotasi), memperbesar (skala), atau meratakan gambar tersebut.
* **Output:** Vektor baru berupa koordinat gambar yang sudah berubah bentuk atau posisi.

Dalam komputasi, operasi transformasi ini dilakukan dengan cara **mengalikan sebuah matriks dengan vektor input** tersebut ($T(x) = Ax$).

## 2. Syarat Mutlak Transformasi "Linear"

Tidak semua fungsi atau pemetaan bisa disebut sebagai transformasi linear. Sebuah fungsi transformasi $T: \mathbb{V} \rightarrow \mathbb{W}$ baru bisa dinyatakan **Linear** jika dan hanya jika memenuhi dua syarat (aksioma) utama berikut:

### Syarat A: Sifat Aditif (Penjumlahan)
Jika kita menjumlahkan dua vektor terlebih dahulu baru ditransformasikan, hasilnya harus sama dengan jika kedua vektor tersebut ditransformasikan masing-masing baru dijumlahkan.
$$T(u + v) = T(u) + T(v)$$

### Syarat B: Sifat Homogenitas (Perkalian Skalar)
Jika sebuah vektor dikalikan dengan sebuah angka skalar $c$ terlebih dahulu baru ditransformasikan, hasilnya harus sama dengan jika vektor tersebut ditransformasikan dulu baru dikalikan dengan skalar $c$.
$$T(cu) = c \cdot T(u)$$

*Catatan: Jika sebuah fungsi transformasi gagal memenuhi salah satu saja dari kedua syarat di atas, maka transformasi tersebut dikategorikan sebagai transformasi non-linear.*

## 3. Ciri Visual Transformasi Linear

Jika diterapkan pada koordinat grafik atau ruang vektor, transformasi linear memiliki dua ciri khas visual yang konsisten:
1. **Garis lurus harus tetap menjadi garis lurus** setelah ditransformasikan (tidak boleh melengkung atau patah).
2. **Titik pusat atau titik asal $(0,0)$ tidak boleh berpindah posisi** (tetap diam di pusat).
