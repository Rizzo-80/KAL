## A. Soal Determinan Matriks 3x3

### **Soal 1**
Hitung determinan dari matriks berikut:
$$A = \begin{bmatrix} 2 & 1 & 3 \\ 0 & 4 & 5 \\ 1 & 2 & 1 \end{bmatrix}$$

#### **Langkah Perhitungan Lengkap:**
Menggunakan metode ekspansi kofaktor sepanjang baris pertama ($i = 1$) dengan rumus umum:
$$\det(A) = \sum_{j=1}^{3} (-1)^{1+j} \cdot a_{1j} \cdot \det(A_{1j})$$

Jika dijabarkan ke dalam bentuk elemen matriks kofaktornya ($C_{1j}$), persamaannya menjadi:
$$\det(A) = (-1)^{1+1} \cdot a_{11} \cdot C_{11} + (-1)^{1+2} \cdot a_{12} \cdot C_{12} + (-1)^{1+3} \cdot a_{13} \cdot C_{13}$$

1. **Substitusi elemen baris pertama dan sub-matriks minornya:**
   * Elemen $a_{11} = 2$, matriks minor didapat dengan menutup baris 1 kolom 1.
   * Elemen $a_{12} = 1$, matriks minor didapat dengan menutup baris 1 kolom 2.
   * Elemen $a_{13} = 3$, matriks minor didapat dengan menutup baris 1 kolom 3.
   
   $$\det(A) = (-1)^{2} \cdot 2 \cdot \begin{vmatrix} 4 & 5 \\ 2 & 1 \end{vmatrix} + (-1)^{3} \cdot 1 \cdot \begin{vmatrix} 0 & 5 \\ 1 & 1 \end{vmatrix} + (-1)^{4} \cdot 3 \cdot \begin{vmatrix} 0 & 4 \\ 1 & 2 \end{vmatrix}$$

2. **Menghitung nilai tanda pangkat $(-1)$ dan determinan matriks $2 \times 2$ (Perkalian Silang Utama - Samping):**
   * $(-1)^2 = 1$ dan $\begin{vmatrix} 4 & 5 \\ 2 & 1 \end{vmatrix} = (4 \times 1) - (5 \times 2) = 4 - 10$
   * $(-1)^3 = -1$ dan $\begin{vmatrix} 0 & 5 \\ 1 & 1 \end{vmatrix} = (0 \times 1) - (5 \times 1) = 0 - 5$
   * $(-1)^4 = 1$ dan $\begin{vmatrix} 0 & 4 \\ 1 & 2 \end{vmatrix} = (0 \times 2) - (4 \times 1) = 0 - 4$

3. **Menggabungkan hasil perkalian:**
   $$\det(A) = (1) \cdot 2 \cdot (4 - 10) - (-1) \cdot (0 - 5) + (1) \cdot 3 \cdot (0 - 4)$$
   $$\det(A) = 2 \cdot (-6) - 1 \cdot (-5) + 3 \cdot (-4)$$
   $$\det(A) = -12 + 5 - 12$$
   $$\det(A) = -19$$

**Jadi, determinan dari matriks diatas adalah $-19$.**

### **Soal 3**
Jika $C = \begin{bmatrix} 1 & 2 & 3 \\ 2 & 4 & 6 \\ 1 & 1 & 1 \end{bmatrix}$, tentukan $\det(C)$ dan jelaskan apakah matriks tersebut singular atau tidak.

#### **Langkah Perhitungan Lengkap:**
Menggunakan rumus aturan ekspansi baris pertama:
$$\text{Rumus: } \det(C) = \sum_{j=1}^{3} (-1)^{1+j} \cdot a_{1j} \cdot \det(C_{1j})$$
$$\det(C) = (-1)^{1+1} \cdot a_{11} \cdot C_{11} - (-1)^{1+2} \cdot a_{12} \cdot C_{12} + (-1)^{1+3} \cdot a_{13} \cdot C_{13}$$

1. **Substitusi nilai elemen baris pertama ($1, 2, 3$) dan matriks minornya:**
   $$\det(C) = (-1)^{2} \cdot 1 \cdot \begin{vmatrix} 4 & 6 \\ 1 & 1 \end{vmatrix} - (-1)^{3} \cdot 2 \cdot \begin{vmatrix} 2 & 6 \\ 1 & 1 \end{vmatrix} + (-1)^{4} \cdot 3 \cdot \begin{vmatrix} 2 & 4 \\ 1 & 1 \end{vmatrix}$$

2. **Menghitung nilai determinan kofaktor diagonal:**
   * Kalikan silang bagian satu: $(4 \times 1) - (6 \times 1) = 4 - 6$
   * Kalikan silang bagian dua: $(2 \times 1) - (6 \times 1) = 2 - 6$
   * Kalikan silang bagian tiga: $(2 \times 1) - (4 \times 1) = 2 - 4$

3. **Akumulasi total nilai numerik:**
   $$\det(C) = (1) \cdot 1 \cdot (4 - 6) - (-1) \cdot 2 \cdot (2 - 6) + (1) \cdot 3 \cdot (2 - 4)$$
   $$\det(C) = 1 \cdot (-2) - 2 \cdot (-4) + 3 \cdot (-2)$$
   $$\det(C) = -2 + 8 - 6$$
   $$\det(C) = 0$$

#### **Penjelasan Analisis Sifat Matriks:**
Setelah mendapatkan nilai $\det(C) = 0$, berdasarkan definisi dasar aljabar matriks, matriks $C$ **terbukti merupakan Matriks Singular**. Hal ini dikarenakan syarat mutlak sebuah matriks disebut singular adalah nilai determinannya yang bernilai nol, yang berakibat matriks ini sama sekali tidak memiliki matriks balikan atau invers.

## B. Soal Dekomposisi Matriks (LU Decomposition)

### **Soal 4**
Lakukan dekomposisi LU terhadap matriks $A = \begin{bmatrix} 2 & 4 & 2 \\ 1 & 5 & 2 \\ 1 & 2 & 4 \end{bmatrix}$ dengan bentuk $A = LU$.

#### **Langkah Perhitungan Detail (Sistem Persamaan Komponen):**
Kita tahu bentuk umum perkalian $L \times U$ adalah:
$$\begin{bmatrix} 2 & 4 & 2 \\ 1 & 5 & 2 \\ 1 & 2 & 4 \end{bmatrix} = \begin{bmatrix} 1 & 0 & 0 \\ l_{21} & 1 & 0 \\ l_{31} & l_{32} & 1 \end{bmatrix} \begin{bmatrix} u_{11} & u_{12} & u_{13} \\ 0 & u_{22} & u_{23} \\ 0 & 0 & u_{33} \end{bmatrix}$$

Mari kita cari nilainya baris demi baris:

1. **Baris 1 Matriks $U$ (Hasil kali Baris 1 $L$ dengan seluruh kolom $U$):**
   * $u_{11} = a_{11} = 2$
   * $u_{12} = a_{12} = 4$
   * $u_{13} = a_{13} = 2$

2. **Kolom 1 Matriks $L$ (Hasil kali Baris 2 & 3 $L$ dengan Kolom 1 $U$):**
   * $l_{21} \times u_{11} = a_{21} \implies l_{21} \times 2 = 1 \implies l_{21} = \frac{1}{2} = 0,5$
   * $l_{31} \times u_{11} = a_{31} \implies l_{31} \times 2 = 1 \implies l_{31} = \frac{1}{2} = 0,5$

3. **Baris 2 Matriks $U$ (Hasil kali Baris 2 $L$ dengan Kolom 2 & 3 $U$):**
   * $(l_{21} \times u_{12}) + u_{22} = a_{22} \implies (0,5 \times 4) + u_{22} = 5 \implies 2 + u_{22} = 5 \implies u_{22} = 3$
   * $(l_{21} \times u_{13}) + u_{23} = a_{23} \implies (0,5 \times 2) + u_{23} = 2 \implies 1 + u_{23} = 2 \implies u_{23} = 1$

4. **Kolom 2 Matriks $L$:**
   * $(l_{31} \times u_{12}) + (l_{32} \times u_{22}) = a_{32}$
   * $(0,5 \times 4) + (l_{32} \times 3) = 2$
   * $2 + 3l_{32} = 2 \implies 3l_{32} = 0 \implies l_{32} = 0$

5. **Baris 3 Matriks $U$:**
   * $(l_{31} \times u_{13}) + (l_{32} \times u_{23}) + u_{33} = a_{33}$
   * $(0,5 \times 2) + (0 \times 1) + u_{33} = 4$
   * $1 + 0 + u_{33} = 4 \implies u_{33} = 3$

**Jawaban Akhir Soal 4:**
$$L = \begin{bmatrix} 1 & 0 & 0 \\ 0,5 & 1 & 0 \\ 0,5 & 0 & 1 \end{bmatrix}, \quad U = \begin{bmatrix} 2 & 4 & 2 \\ 0 & 3 & 1 \\ 0 & 0 & 3 \end{bmatrix}$$

### **Soal 5**
Tentukan matriks $L$ dan $U$ dari $B = \begin{bmatrix} 1 & 2 & 1 \\ 2 & 5 & 3 \\ 4 & 10 & 8 \end{bmatrix}$ menggunakan metode eliminasi Gauss.

#### **Langkah Perhitungan Lengkap (Operasi Baris Elementer):**
Kita lakukan reduksi baris pada matriks $B$ untuk menjadikannya matriks segitiga atas ($U$), lalu setiap bilangan pengali kofaktornya dicatat untuk menyusun matriks segitiga bawah ($L$).

1. **Mengeliminasi Elemen Kolom 1 (Baris 2 dan Baris 3):**
   * **Untuk Baris 2 ($R_2$):** Kita ingin mengubah angka $2$ menjadi $0$ menggunakan patokan elemen diagonal baris 1 ($1$). 
     * Rumus operasi baris: $R_2 - 2R_1$
     * Nilai pengali kofaktor dicatat sebagai: $l_{21} = 2$
     * Proses hitung elemen demi elemen:
       $$\begin{aligned}
       R_2 \text{ baru} &= \begin{bmatrix} 2 & 5 & 3 \end{bmatrix} - 2 \times \begin{bmatrix} 1 & 2 & 1 \end{bmatrix} \\
       &= \begin{bmatrix} (2 - 2\cdot1) & (5 - 2\cdot2) & (3 - 2\cdot1) \end{bmatrix} \\
       &= \begin{bmatrix} 0 & 1 & 1 \end{bmatrix}
       \end{aligned}$$

   * **Untuk Baris 3 ($R_3$):** Kita ingin mengubah angka $4$ menjadi $0$ menggunakan patokan elemen diagonal baris 1 ($1$).
     * Rumus operasi baris: $R_3 - 4R_1$
     * Nilai pengali kofaktor dicatat sebagai: $l_{31} = 4$
     * Proses hitung elemen demi elemen:
       $$\begin{aligned}
       R_3 \text{ baru} &= \begin{bmatrix} 4 & 10 & 8 \end{bmatrix} - 4 \times \begin{bmatrix} 1 & 2 & 1 \end{bmatrix} \\
       &= \begin{bmatrix} (4 - 4\cdot1) & (10 - 4\cdot2) & (8 - 4\cdot1) \end{bmatrix} \\
       &= \begin{bmatrix} 0 & 2 & 4 \end{bmatrix}
       \end{aligned}$$
   
   Matriks hasil transformasi tahap pertama:
   $$\begin{bmatrix} 1 & 2 & 1 \\ 0 & 1 & 1 \\ 0 & 2 & 4 \end{bmatrix}$$

2. **Mengeliminasi Elemen Kolom 2 (Baris 3):**
   * Kita ingin mengubah angka $2$ (pada baris 3 kolom 2) menjadi $0$ menggunakan patokan elemen diagonal baris 2 yang baru ($1$).
     * Rumus operasi baris: $R_3 - 2R_2$
     * Nilai pengali kofaktor dicatat sebagai: $l_{32} = 2$
     * Proses hitung elemen demi elemen:
       $$\begin{aligned}
       R_3 \text{ baru} &= \begin{bmatrix} 0 & 2 & 4 \end{bmatrix} - 2 \times \begin{bmatrix} 0 & 1 & 1 \end{bmatrix} \\
       &= \begin{bmatrix} (0 - 2\cdot0) & (2 - 2\cdot1) & (4 - 2\cdot1) \end{bmatrix} \\
       &= \begin{bmatrix} 0 & 0 & 2 \end{bmatrix}
       \end{aligned}$$

   Matriks sepenuhnya telah tereduksi menjadi Matriks Segitiga Atas ($U$):
   $$U = \begin{bmatrix} 1 & 2 & 1 \\ 0 & 1 & 1 \\ 0 & 0 & 2 \end{bmatrix}$$

3. **Menyusun Matriks Segitiga Bawah ($L$):**
   Gunakan struktur diagonal utama bernilai 1, lalu isi posisi kosong di bawah diagonal dengan nilai kofaktor pengali yang telah kita catat ($l_{21} = 2$, $l_{31} = 4$, $l_{32} = 2$):
   $$L = \begin{bmatrix} 1 & 0 & 0 \\ l_{21} & 1 & 0 \\ l_{31} & l_{32} & 1 \end{bmatrix} \implies L = \begin{bmatrix} 1 & 0 & 0 \\ 2 & 1 & 0 \\ 4 & 2 & 1 \end{bmatrix}$$

**Jawaban Akhir Soal 5:**
$$L = \begin{bmatrix} 1 & 0 & 0 \\ 2 & 1 & 0 \\ 4 & 2 & 1 \end{bmatrix}, \quad U = \begin{bmatrix} 1 & 2 & 1 \\ 0 & 1 & 1 \\ 0 & 0 & 2 \end{bmatrix}$$

## C. Soal Invers Matriks 3x3

### **Soal 7**

#### **Langkah Perhitungan Lengkap (Metode Adjoin):**
Rumus Invers: $A^{-1} = \frac{1}{\det(A)} \cdot \text{Adj}(A)$

**Langkah 1: Menghitung Determinan $\det(A)$**
Ekspansi Baris 1:
$$\det(A) = 1(4 - 3) - 2(0 - 2) + 1(0 - 2) = 1(1) - 2(-2) + 1(-2) = 1 + 4 - 2 = 3$$

**Langkah 2: Mencari Kofaktor Sembilan Elemen ($C_{ij} = (-1)^{i+j} \cdot M_{ij}$)**
* $C_{11} = +\det\begin{bmatrix} 1 & 1 \\ 3 & 4 \end{bmatrix} = (4-3) = 1$
* $C_{12} = -\det\begin{bmatrix} 0 & 1 \\ 2 & 4 \end{bmatrix} = -(0-2) = 2$
* $C_{13} = +\det\begin{bmatrix} 0 & 1 \\ 2 & 3 \end{bmatrix} = (0-2) = -2$
* $C_{21} = -\det\begin{bmatrix} 2 & 1 \\ 3 & 4 \end{bmatrix} = -(8-3) = -5$
* $C_{22} = +\det\begin{bmatrix} 1 & 1 \\ 2 & 4 \end{bmatrix} = (4-2) = 2$
* $C_{23} = -\det\begin{bmatrix} 1 & 2 \\ 2 & 3 \end{bmatrix} = -(3-4) = 1$
* $C_{31} = +\det\begin{bmatrix} 2 & 1 \\ 1 & 1 \end{bmatrix} = (2-1) = 1$
* $C_{32} = -\det\begin{bmatrix} 1 & 1 \\ 0 & 1 \end{bmatrix} = -(1-0) = -1$
* $C_{33} = +\det\begin{bmatrix} 1 & 2 \\ 0 & 1 \end{bmatrix} = (1-0) = 1$

Maka susunan Matriks Kofaktor:
$$\text{Kofaktor}(A) = \begin{bmatrix} 1 & 2 & -2 \\ -5 & 2 & 1 \\ 1 & -1 & 1 \end{bmatrix}$$

**Langkah 3: Membuat Matriks Adjoin ($\text{Adj}(A) = \text{Kofaktor}(A)^T$)**
Balik baris menjadi kolom pada matriks kofaktor:
$$\text{Adj}(A) = \begin{bmatrix} 1 & -5 & 1 \\ 2 & 2 & -1 \\ -2 & 1 & 1 \end{bmatrix}$$

**Langkah 4: Menghitung Invers Akhir**
$$A^{-1} = \frac{1}{3} \cdot \begin{bmatrix} 1 & -5 & 1 \\ 2 & 2 & -1 \\ -2 & 1 & 1 \end{bmatrix}$$

**Jawaban Akhir Soal 7:**

### **Soal 9**
