### Definition. Matriks
Sebuah **matriks** adalah susunan bilangan berbentuk persegi panjang. Matriks dengan $m$ baris dan $n$ kolom dikatakan memiliki ukuran (atau dimensi) $m \times n$.

### Definition. Matriks Augmentasi
Misalkan $L$ adalah sistem linear. Matriks augmentasi yang terkait dengan $L$ adalah matriks:

$$\begin{bmatrix}
a_{11} & a_{12} & \dots & a_{1n} & \big| & b_1 \\
a_{21} & a_{22} & \dots & a_{2n} & \big| & b_2 \\
\vdots & \vdots & \ddots & \vdots & \big| & \vdots \\
a_{m1} & a_{m2} & \dots & a_{mn} & \big| & b_m
\end{bmatrix}$$

---

## Matriks Bentuk Eselon Baris

Berikut adalah formulasi untuk sistem linear “sederhana”; yaitu sistem yang matriks augmentasinya berada dalam bentuk eselon baris, seperti yang dijelaskan di bawah.

### Definition. Bentuk Eselon Baris
**Baris nol** dari suatu matriks adalah baris yang semua entrinya sama dengan nol; **baris tak-nol** adalah baris yang memuat setidaknya satu entri tak-nol.

Suatu matriks berada dalam **bentuk eselon baris** jika kondisi-kondisi berikut terpenuhi:
1. Pada sembarang baris tak-nol, entri tak-nol pertama (yaitu, paling kiri) sama dengan satu. Satu utama (pivot) dari suatu matriks adalah entri dari suatu baris yang sama dengan satu, dan merupakan entri tak-nol pertama dari baris tersebut.
2. Semua baris nol dikelompokkan bersama di bagian bawah matriks.
3. Diberikan dua baris tak-nol sembarang dalam matriks, satu utama (pivot) dari baris yang lebih bawah terletak di sebelah kanan satu utama dari baris di atasnya.

Suatu matriks berada dalam **bentuk eselon baris tereduksi** jika selain memenuhi kondisi (1)-(3) juga memenuhi kondisi berikut:
4. Setiap kolom matriks yang memuat satu utama pada salah satu barisnya, maka itu adalah satu-satunya entri tak-nol dari kolom tersebut.

> **Aturan Praktis Pemeriksaan:**
> * Pertama verifikasi apakah semua baris nol berada di bagian bawah.
> * Untuk setiap baris tak-nol, tentukan apakah entri tak-nol pertama adalah 1, dan tandai kotak pada pivot tersebut.
> * Pastikan kotak-kotak tersebut membentuk pola tangga.
> * *(Untuk eselon baris tereduksi)* Tentukan apakah setiap kolom selain pivot memiliki nilai 0.

## Eliminasi Gaussian

Prosedur sistematis yang disebut **eliminasi Gaussian** memungkinkan kita mereduksi sistem linear $L$ menjadi sistem $L'$ dalam bentuk eselon baris dengan melakukan **Operasi Baris Elementer (OBE)**.

### Definition. Operasi Baris Elementer pada Matriks
Misalkan $A$ adalah matriks yang diberikan, dan nyatakan dengan $R_i$ baris ke-$i$ dari $A$. Terdapat tiga jenis operasi matriks dasar:

1. **Perkalian Skalar:** Kalikan sebuah baris dengan bilangan tak-nol $k$ ($R_i \leftarrow k R_i$).
2. **Pertukaran Baris:** Tukar posisi dua baris dari $A$ ($R_i \leftrightarrow R_j$).
3. **Penjumlahan Baris:** Tambahkan kelipatan satu baris ke baris lain ($R_i \leftarrow R_i + k R_j$).

Proses mengubah matriks menggunakan operasi baris elementer disebut **reduksi baris**. Dua matriks dikatakan **ekuivalen baris** jika salah satunya diperoleh dari yang lain melalui serangkaian OBE.

### Definition. Algoritma Eliminasi Gaussian
* **Langkah 1:** Temukan kolom tak-nol paling kiri dan lakukan pertukaran baris untuk memindahkan baris dengan entri tak-nol ini ke bagian atas matriks.
* **Langkah 2:** Skala baris teratas yang baru untuk menghasilkan satu utama pada baris tersebut. Sebut baris baru ini $R_1$.
* **Langkah 3:** Untuk setiap baris $R_k$ di bawah $R_1$, lakukan operasi baris berbentuk $R_k \leftarrow R_k - c R_1$ untuk mengubah semua entri di bawah satu utama menjadi nol.
* **Langkah 4:** Mulai lagi dengan Langkah 1 yang diterapkan pada matriks yang terdiri dari semua baris di bawah $R_1$. Lanjutkan hingga matriks berada dalam bentuk eselon baris.

### Definition. Algoritma Eliminasi Gauss-Jordan
* **Langkah 1-4:** Terapkan eliminasi Gaussian untuk mengubah matriks menjadi bentuk eselon baris.
* **Langkah 5:** Temukan kolom paling kanan dari matriks yang memuat satu utama. Misalkan $R_i$ adalah baris yang memuat satu utama ini. Untuk setiap baris $R_k$ di atas $R_i$, lakukan operasi baris berbentuk $R_k \leftarrow R_k - c R_i$ untuk mengganti semua entri di atas satu utama dengan nol.
* **Langkah 6:** Mulai lagi dengan Langkah 5 untuk diterapkan pada kolom berikutnya di sebelah kiri yang memuat satu utama hingga matriks berada dalam bentuk eselon baris tereduksi.

# Studi Kasus: Eliminasi Gaussian SPL 4x4

Halaman ini membahas penyelesaian Sistem Persamaan Linear (SPL) dengan ordo $4 \times 4$ menggunakan metode Eliminasi Gaussian hingga menghasilkan bentuk eselon baris tereduksi.

## Representasi Sistem Persamaan Linear dan Matriks Augmented

Misalkan kita memiliki sistem persamaan linear $4 \times 4$ sebagai berikut:

1. $x_1 + x_2 + x_3 + x_4 = 10$
2. $2x_1 + 3x_2 + 2x_3 + 3x_4 = 23$
3. $x_1 + 2x_2 + 3x_3 + 2x_4 = 19$
4. $x_1 + x_2 + x_3 + 2x_4 = 12$

Jika direpresentasikan ke dalam bentuk **Matriks Augmented (Matriks Diperbesar)**, maka susunannya adalah:

$$\begin{bmatrix}
1 & 1 & 1 & 1 & \big| & 10 \\
2 & 3 & 2 & 3 & \big| & 23 \\
1 & 2 & 3 & 2 & \big| & 19 \\
1 & 1 & 1 & 2 & \big| & 12
\end{bmatrix}$$

---

## Proses Transformasi Operasi Baris Elementer (OBE)

Tujuan kita adalah mengubah matriks di atas menjadi matriks eselon baris tereduksi menggunakan Operasi Baris Elementer (OBE).

### Tahap 1: Analisis Pivot Kolom Pertama dan Eliminasi Baris di Bawahnya
Pada baris pertama kolom pertama ($a_{11}$), kita sudah mendapatkan nilai pivot utama bernilai **1**. Langkah selanjutnya adalah menolkan semua elemen di bawah pivot tersebut (Baris 2 sampai Baris 4).

* **Operasi OBE:**
  * $R_2 \leftarrow R_2 - 2R_1$
  * $R_3 \leftarrow R_3 - R_1$
  * $R_4 \leftarrow R_4 - R_1$

Hasil transformasi matriks:
$$\begin{bmatrix}
1 & 1 & 1 & 1 & \big| & 10 \\
0 & 1 & 0 & 1 & \big| & 3 \\
0 & 1 & 2 & 1 & \big| & 9 \\
0 & 0 & 0 & 1 & \big| & 2
\end{bmatrix}$$

### Tahap 2: Analisis Kolom Kedua dan Ketiga
Perhatikan Baris 2 kolom kedua ($a_{22}$), nilainya sudah otomatis bernilai **1** (pivot kedua). Sekarang kita gunakan baris kedua ini untuk menolkan elemen di Baris 3 kolom kedua ($a_{32}$).

* **Operasi OBE:**
  * $R_3 \leftarrow R_3 - R_2$

Hasil transformasi matriks:
$$\begin{bmatrix}
1 & 1 & 1 & 1 & \big| & 10 \\
0 & 1 & 0 & 1 & \big| & 3 \\
0 & 0 & 2 & 0 & \big| & 6 \\
0 & 0 & 0 & 1 & \big| & 2
\end{bmatrix}$$

Selanjutnya, buat elemen pada Baris 3 kolom ketiga ($a_{33}$) menjadi 1 utama (pivot ketiga) dengan membagi Baris 3 dengan angka 2.
* **Operasi OBE:**
  * $R_3 \leftarrow \frac{1}{2}R_3$

Hasil transformasi matriks:
$$\begin{bmatrix}
1 & 1 & 1 & 1 & \big| & 10 \\
0 & 1 & 0 & 1 & \big| & 3 \\
0 & 0 & 1 & 0 & \big| & 3 \\
0 & 0 & 0 & 1 & \big| & 2
\end{bmatrix}$$

*(Matriks di atas sekarang sudah sukses membentuk format Eselon Baris karena seluruh baris terbawah memiliki angka 1 utama di sebelah kanan baris di atasnya).*

### Tahap 3: Eliminasi ke Atas (Menuju Eselon Baris Tereduksi)
Untuk membuat matriks identitas penuh di sisi kiri, kita kurangi elemen Baris 1 dan Baris 2 dengan elemen baris di bawahnya agar kolom di atas pivot bernilai 0.

* **Operasi OBE:**
  * $R_2 \leftarrow R_2 - R_4$ (Menolkan $a_{24}$)
  * $R_1 \leftarrow R_1 - R_4$ (Menolkan $a_{14}$)
  * $R_1 \leftarrow R_1 - R_3$ (Menolkan $a_{13}$)
  * $R_1 \leftarrow R_1 - R_2$ (Menolkan $a_{12}$)

Hasil akhir transformasi matriks:
$$\begin{bmatrix}
1 & 0 & 0 & 0 & \big| & 4 \\
0 & 1 & 0 & 0 & \big| & 1 \\
0 & 0 & 1 & 0 & \big| & 3 \\
0 & 0 & 0 & 1 & \big| & 2
\end{bmatrix}$$

## Kesimpulan Himpunan Penyelesaian

Berdasarkan hasil matriks eselon baris tereduksi di atas, nilai dari masing-masing variabel langsung ditemukan tanpa perlu substitusi balik:

* $x_1 = 4$
* $x_2 = 1$
* $x_3 = 3$
* $x_4 = 2$

## Validasi Logika Menggunakan Kode Python (NumPy)

Berikut adalah kode Python untuk memvalidasi kebenaran nilai variabel dari sistem persamaan linear $4 \times 4$ di atas:



````python
import numpy as np

A = np.array([
    [1, 1, 1, 1],
    [2, 3, 2, 3],
    [1, 2, 3, 2],
    [1, 1, 1, 2]
])

B = np.array([10, 23, 19, 12])

X = np.linalg.solve(A, B)

print("Hasil Validasi Variabel (x1 sampai x4):")
print(f"x1 = {X[0]}")
print(f"x2 = {X[1]}")
print(f"x3 = {X[2]}")
print(f"x4 = {X[3]}")
````
