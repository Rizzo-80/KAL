# Apa itu Komputasi Aljabar Linear?

**Komputasi Aljabar Linear** adalah cabang ilmu yang menggabungkan teori matematika Aljabar Linear dengan algoritma komputer untuk menyelesaikan perhitungan matriks dan vektor dalam skala yang sangat besar.

Jika dalam kelas teori kita menghitung matriks ukuran $2 \times 2$ atau $3 \times 3$ secara manual menggunakan kertas dan pulpen, maka dalam **Komputasi** Aljabar Linear, kita menggunakan bahasa pemrograman (seperti **Python**) untuk menghitung matriks berukuran ribuan hingga jutaan baris dan kolom secara instan.


## Mengapa Komputasi Aljabar Linear Penting?

Hampir semua teknologi modern di bidang Sains Data dan Informatika dibangun di atas pondasi Aljabar Linear yang dikomputerisasi. Berikut adalah beberapa alasan utamanya:

* **Pondasi Data Science & Machine Learning:** Data seperti gambar, teks, dan suara akan diubah oleh komputer menjadi tumpukan angka berbentuk matriks. Algoritma seperti *K-Nearest Neighbors* (K-NN) atau *Neural Networks* bekerja dengan cara mengalikan matriks-matriks tersebut.
* **Pemrosesan Gambar (Image Processing):** Sebuah foto digital sebenarnya adalah matriks besar, di mana setiap elemen matriksnya menyimpan nilai warna (pixel). Proses filter, rotasi, atau kompresi gambar semuanya menggunakan operasi matriks.
* **Sistem Persamaan Linear Skala Besar:** Digunakan untuk memprediksi data, analisis jaringan, hingga grafika komputer 3D (seperti pada *game engine*).


## Topik Utama yang Dipelajari

Dalam mata kuliah ini, fokus utama kita adalah mengimplementasikan teori-teori berikut ke dalam kode program:

1. **Sistem Persamaan Linear (SPL):** Menyelesaikan nilai variabel tak diketahui menggunakan metode seperti Eliminasi Gauss atau Gauss-Jordan.
2. **Operasi Matriks & Vektor:** Perkalian matriks, translasi, dan transformasi linear.
3. **Determinan dan Invers Matriks:** Menghitung nilai karakteristik matriks dan mencari kebalikan matriks untuk menyelesaikan persamaan matematika.
4. **Nilai Eigen dan Vektor Eigen (Eigenvalues & Eigenvectors):** Digunakan untuk reduksi dimensi data dan algoritma rekomendasi.


> **Kesimpulan:** > Komputasi Aljabar Linear bukan sekadar menghitung angka, melainkan bagaimana kita menerjemahkan data dunia nyata menjadi bentuk matriks, lalu menyuruh komputer menyelesaikannya dengan cepat dan akurat.

## Contoh Soal Aljabar Linear (Ordo 2x2)

Berikut adalah contoh soal Sistem Persamaan Linear (SPL) dua variabel yang biasa diselesaikan menggunakan operasi matriks ordo $2 \times 2$.

### Soal:
Tentukan nilai $x$ dan $y$ dari sistem persamaan berikut:
1. $2x + y = 5$
2. $x + 3y = 5$

---

### 1. Representasi Bentuk Matriks
Persamaan linear di atas jika diubah ke dalam bentuk matriks ($Ax = B$) menjadi:

$$\begin{pmatrix} 2 & 1 \\ 1 & 3 \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} 5 \\ 5 \end{pmatrix}$$

* **Matriks $A$:** $\begin{pmatrix} 2 & 1 \\ 1 & 3 \end{pmatrix}$
* **Vektor $B$:** $\begin{pmatrix} 5 \\ 5 \end{pmatrix}$

---

### 2. Cara Visualisasi di GeoGebra

Untuk membuktikan jawaban dari soal ini menggunakan GeoGebra, kamu cukup menggambar kedua persamaan tersebut menjadi garis lurus:

* **Hasil Titik Potong:** Kedua garis akan berpotongan persis di titik **$(2, 1)$**.
* **Kesimpulan:** Nilai $x = 2$ dan $y = 1$ adalah solusi akhir dari soal Aljabar Linear ini.

## 4. Implementasi Menggunakan Kode Python (NumPy)

Untuk menyelesaikan Sistem Persamaan Linear (SPL) dari soal di atas secara otomatis, kita bisa menggunakan bahasa pemrograman Python dengan bantuan *library* **NumPy**. NumPy menyediakan fungsi khusus bernama `np.linalg.solve` yang dapat mencari nilai variabel $x$ dan $y$ secara instan menggunakan operasi matriks.

Berikut adalah kode program Python untuk menyelesaikan kasus ordo $2 \times 2$ tersebut:


```python
import numpy as np

A = np.array([[2, 1], 
              [1, 3]])

B = np.array([5, 5])

X = np.linalg.solve(A, B)

print("Matriks A:")
print(A)
print("\nVektor B:")
print(B)
print("\nHasil Solusi (Nilai x dan y):")
print(f"x = {X[0]}")
print(f"y = {X[1]}")
```
