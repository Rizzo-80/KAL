## Komputasi Determinan dengan Ekspansi Baris dan Kolom

Halaman ini membahas secara mendalam konsep perhitungan nilai determinan matriks persegi menggunakan metode Ekspansi Kofaktor (Laplace), baik melalui pendekatan ekspansi baris maupun ekspansi kolom.

---

## Penjelasan Teori

Metode Ekspansi Kofaktor digunakan untuk menghitung determinan matriks berordo besar (seperti $3 \times 3$ atau lebih) dengan cara memecahnya menjadi sub-matriks yang lebih kecil (minor). Dalam perhitungannya, kita bebas memilih untuk menghitung sepanjang **satu baris** atau **satu kolom** mana saja karena hasil akhirnya akan selalu sama.

Setiap elemen yang dipilih akan dikalikan dengan nilai **Kofaktor ($C_{ij}$)** miliknya, yang dirumuskan sebagai:

$$C_{ij} = (-1)^{i+j} M_{ij}$$

* **Aturan Tanda $(-1)^{i+j}$:** Menghasilkan pola tanda plus-minus yang berselang-seling seperti papan catur. Untuk matriks $3 \times 3$, polanya adalah:
  $$\begin{bmatrix}
  + & - & + \\
  - & + & - \\
  + & - & +
  \end{bmatrix}$$
* **Minor ($M_{ij}$):** Nilai determinan dari sub-matriks $2 \times 2$ yang tersisa setelah kamu menghapus baris ke-$i$ dan kolom ke-$j$.

### 1. Pendekatan Ekspansi Baris
Ekspansi baris dilakukan dengan cara berjalan mendatar sepanjang elemen baris yang dipilih. Jika kita memilih **Baris Pertama** ($a_{11}, a_{12}, a_{13}$), maka rumus determinannya:

$$\det(A) = a_{11}C_{11} + a_{12}C_{12} + a_{13}C_{13}$$

### 2. Pendekatan Ekspansi Kolom
Ekspansi kolom dilakukan dengan cara berjalan menurun sepanjang elemen kolom yang dipilih. Jika kita memilih **Kolom Pertama** ($a_{11}, a_{21}, a_{31}$), maka rumus determinannya:

$$\det(A) = a_{11}C_{11} + a_{21}C_{21} + a_{31}C_{31}$$

*Tips: Ekspansi kolom sangat disarankan jika ada kolom yang memiliki banyak angka nol ($0$), karena akan memangkas banyak waktu hitungan.*

---

## Contoh Soal dan Penyelesaian Manual

Misalkan kita memiliki matriks persegi $A$ berordo $3 \times 3$ sebagai berikut:

$$A = \begin{bmatrix}
3 & 2 & -1 \\
0 & 4 & 1 \\
2 & 1 & 5
\end{bmatrix}$$

Mari kita buktikan bahwa ekspansi baris pertama dan ekspansi kolom pertama menghasilkan nilai determinan yang sama.

### Alternatif A: Menghitung Menggunakan Ekspansi Baris Pertama

$$\det(A) = a_{11}C_{11} + a_{12}C_{12} + a_{13}C_{13}$$

* **Komponen Elemen $a_{11} = 3$ (Tanda $+$):**
  $$C_{11} = 1 \cdot \det \begin{bmatrix} 4 & 1 \\ 1 & 5 \end{bmatrix} = (4 \cdot 5) - (1 \cdot 1) = 20 - 1 = 19$$

* **Komponen Elemen $a_{12} = 2$ (Tanda $-$):**
  $$C_{12} = -1 \cdot \det \begin{bmatrix} 0 & 1 \\ 2 & 5 \end{bmatrix} = -1 \cdot ((0 \cdot 5) - (1 \cdot 2)) = -1 \cdot (-2) = 2$$

* **Komponen Elemen $a_{13} = -1$ (Tanda $+$):**
  $$C_{13} = 1 \cdot \det \begin{bmatrix} 0 & 4 \\ 2 & 1 \end{bmatrix} = (0 \cdot 1) - (4 \cdot 2) = 0 - 8 = -8$$

**Hitung Total Determinan:**
$$\det(A) = 3(19) + 2(2) + (-1)(-8)$$
$$\det(A) = 57 + 4 + 8 = 69$$

---

### Alternatif B: Menghitung Menggunakan Ekspansi Kolom Pertama

$$\det(A) = a_{11}C_{11} + a_{21}C_{21} + a_{31}C_{31}$$

* **Komponen Elemen $a_{11} = 3$ (Tanda $+$):**
  Nilai $C_{11}$ sama dengan perhitungan di atas, yaitu **19**.

* **Komponen Elemen $a_{21} = 0$ (Tanda $-$):**
  Karena elemennya bernilai $0$, kita tidak perlu menghitung minor matriksnya karena hasil perkaliannya pasti nol:
  $$a_{21}C_{21} = 0 \cdot C_{21} = 0$$

* **Komponen Elemen $a_{31} = 2$ (Tanda $+$):**
  $$C_{31} = 1 \cdot \det \begin{bmatrix} 2 & -1 \\ 4 & 1 \end{bmatrix} = (2 \cdot 1) - (-1 \cdot 4) = 2 + 4 = 6$$

**Hitung Total Determinan:**
$$\det(A) = 3(19) - 0 + 2(6)$$
$$\det(A) = 57 - 0 + 12 = 69$$

*Kesimpulan: Terbukti bahwa baik menggunakan ekspansi baris pertama maupun kolom pertama, nilai determinan matriks $A$ tetap bernilai **69**.*

## Implementasi Validasi Menggunakan Python (NumPy)

Kode program di bawah ini menerapkan fungsi bawaan dari pustaka NumPy untuk memverifikasi ketepatan nilai determinan yang telah dihitung secara manual:

```python

import numpy as np

A = np.array([
    [3, 2, -1],
    [0, 4, 1],
    [2, 1, 5]
])

det_A = np.linalg.det(A)

print("Hasil Validasi Komputasi Determinan:")
print(f"det(A) = {round(det_A)}")
```
